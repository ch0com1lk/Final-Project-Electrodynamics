```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq
import warnings
warnings.filterwarnings('ignore')
```


### **1. ECUACIÓN DE ONDA VECTORIAL EN MEDIOS DIELÉCTRICOS**


```python
print("\n" + "="*80)
print("DERIVACIÓN DE LA ECUACIÓN DE ONDA")
print("="*80)

print("""
Partimos de las ecuaciones de Maxwell en un medio lineal, isótropo y homogéneo
sin fuentes (ρ = 0, J = 0):

∇·D = 0                        (Ley de Gauss)
∇·B = 0                        (Ausencia de monopolos magnéticos)
∇×E = -∂B/∂t                   (Ley de Faraday)
∇×H = ∂D/∂t                    (Ley de Ampère-Maxwell)

Con las relaciones constitutivas: D = εE, B = μH

Tomando rotacional de la Ley de Faraday:
∇×(∇×E) = -∂/∂t(∇×B)

Sustituyendo ∇×B = με∂E/∂t:
∇×(∇×E) = -με∂²E/∂t²

Usando la identidad vectorial: ∇×(∇×E) = ∇(∇·E) - ∇²E
y el hecho de que ∇·E = 0 en el medio homogéneo:

RESULTADO: ∇²E - με∂²E/∂t² = 0

Esta es la ecuación de onda vectorial que gobierna la propagación.
""")
```

### **2. VALIDACIÓN: ONDAS PLANAS COMO SOLUCIONES**

```python
print("\n" + "="*80)
print("VALIDACIÓN DE ONDAS PLANAS MONOCROMÁTICAS")
print("="*80)

print("""
Proponemos una solución de onda plana monocromática:

E(r,t) = E₀ exp(i(k·r - ωt))

Aplicando los operadores diferenciales:
∂²E/∂t² = (-iω)²E = -ω²E
∇²E = (i∇)²E = -k²E

Sustituyendo en la ecuación de onda:
-k²E + meω²E = 0

Para soluciones no triviales (E ≠ 0):
RELACIÓN DE DISPERSIÓN: k² = ω²με

Esta es la condición que debe cumplirse para que la onda plana sea solución.
La onda plana es solución válida de la ecuación de onda
""")
```

### **3. POLARIZACIÓN TE: RELACIÓN ENTRE E_y Y H_x**

```python
print("\n" + "="*80)
print("RELACIONES DE POLARIZACIÓN TE (E_y, H_x)")
print("="*80)

print("""
Consideramos ondas TE (Transversas Eléctricas) con:
E = (0, E_y, 0)      Campo eléctrico transversal
H = (H_x, 0, H_z)    Campo magnético

Propagación en el plano xz con dependencia e^(i(k_x·x + k_z·z - ωt))

De la Ley de Faraday: ∇×E = iωμH

Calculando el rotacional:
∇×E = |  i      j      k  |
       | ∂/∂x  ∂/∂y  ∂/∂z |
       |  0     E_y    0  |

∇×E = -∂E_y/∂z î + ∂E_y/∂x k̂

Para ondas planas: ∂/∂z → ik_z, ∂/∂x → ik_x

∇×E = -ik_z·E_y î + ik_x·E_y k̂

Igualando la componente î con iωμH_x:
-ik_z·E_y = iωμ·H_x

RELACIÓN DE IMPEDANCIA:
H_x = -(k_z)/(ωμ) E_y

O equivalentemente:
E_y/H_x = -(ωμ)/k_z

Esta relación acopla las componentes del campo en modos TE.
""")
```

### **4. CONFIGURACIÓN DE LA GUÍA DE ONDA DE TRES CAPAS**

```python
print("\n" + "="*80)
print("GEOMETRÍA Y PARÁMETROS DE LA GUÍA DE ONDA")
print("="*80)

# Parámetros físicos
n1 = 3.3          # Índice de refracción de la cubierta (cladding)
n2 = 3.6          # Índice de refracción del núcleo (core)
d = 2.0           # Espesor de la capa central (micrometros)
wavelength = 0.87 # Longitud de onda (micrometros)

print(f"""
ESTRUCTURA :

    z < 0          z ∈ [0,d]        z > d

    Medium I       Medium II        Medium III
    n₁ = {n1}      n₂ = {n2}        n₁ = {n1}
    (Cladding)     (Core)           (Cladding)

    ~~~~~~~~~~~~~|==================|~~~~~~~~~~~~~
               z=0                z=d

PARÁMETROS:
    Índice de refracción core:     n₂ = {n2}
    Índice de refracción cladding: n₁ = {n1}
    Espesor del núcleo:            d = {d} μm
    Longitud de onda:              λ = {wavelength} μm

MATERIAL: GaAs/AlGaAs
""")

# Cálculos preliminares
k0 = 2 * np.pi / wavelength
NA = np.sqrt(n2**2 - n1**2)
V = k0 * d * NA / 2

print(f"""
PARÁMETROS DERIVADOS:
    k₀ = 2π/λ = {k0:.6f} μm⁻¹
    Apertura Numérica (NA) = √(n₂² - n₁²) = {NA:.6f}
    Número V (normalizado) = k₀·d·NA/2 = {V:.6f}

Número de modos teórico: V/π ≈ {V/np.pi:.2f}

El parámetro V determina cuántos modos pueden propagarse.
Específicamente, el número aproximado de modos TE es:
    N_modos ≈ V/π (para V >> 1)
""")
```

### **5. CONDICIONES FRONTERA**

```python
print("\n" + "="*80)
print("CONDICIONES DE FRONTERA EN LAS INTERFACES")
print("="*80)

print("""
En las interfaces dieléctricas (sin corrientes ni cargas superficiales):

CONTINUIDAD DE COMPONENTES TANGENCIALES:

En z = 0:
    1) E_y(0⁻) = E_y(0⁺)           → Continuidad de E_y
    2) H_x(0⁻) = H_x(0⁺)           → Continuidad de H_x
                                    → Continuidad de ∂E_y/∂z (con μ = cte)

En z = d:
    3) E_y(d⁻) = E_y(d⁺)           → Continuidad de E_y
    4) H_x(d⁻) = H_x(d⁺)           → Continuidad de H_x
                                    → Continuidad de ∂E_y/∂z

Estas cuatro ecuaciones definen un sistema que determina los modos permitidos.
""")
```

### **6. ECUACIÓN CARACTERÍSTICA Y SOLUCIÓN DE MODOS**

```python
print("\n" + "="*80)
print("SOLUCIÓN PARA LOS MODOS GUIADOS")
print("="*80)

def resolver_modos_guiados(n1, n2, d, wavelength, num_modos=4):
    """
    Resuelve la ecuación característica transcendental para modos guiados TE
    en una guía de onda simétrica de tres capas.

    Parámetros:
    -----------
    n1 : float
        Índice de refracción de la cubierta
    n2 : float
        Índice de refracción del núcleo
    d : float
        Espesor del núcleo
    wavelength : float
        Longitud de onda
    num_modos : int
        Número de modos a calcular

    Retorna:
    --------
    lista : Lista con los parámetros de cada modo
    """

    k0 = 2 * np.pi / wavelength
    NA = np.sqrt(n2**2 - n1**2)
    V = k0 * d * NA / 2

    modos = []

    # Ecuación característica para modos pares (simétricos):
    # u·tan(u) = w, donde u = h·d/2, w = q·d/2, u² + w² = V²
    def ec_par(u):
        if u <= 0 or u >= V - 1e-8:
            return np.nan
        w = np.sqrt(V**2 - u**2)
        return u * np.tan(u) - w

    # Ecuación característica para modos impares (antisimétricos):
    # -u·cot(u) = w
    def ec_impar(u):
        if u <= 0 or u >= V - 1e-8:
            return np.nan
        w = np.sqrt(V**2 - u**2)
        tan_u = np.tan(u)
        if np.abs(tan_u) < 1e-10:
            return np.nan
        return -u / tan_u - w

    print(f"\nBÚSQUEDA DE RAÍCES:")
    print(f"{'Modo':<8} {'Tipo':<15} {'u':<12} {'w':<12} {'h (μm⁻¹)':<14} {'q (μm⁻¹)':<14} {'β (μm⁻¹)':<14} {'n_eff':<10}")
    print("-" * 110)

    for m in range(num_modos):
        es_par = (m % 2 == 0)

        # Búsqueda en el intervalo apropiado
        limite_inf = m * (np.pi / 2) + 1e-4
        limite_sup = (m + 1) * (np.pi / 2) - 1e-4

        if limite_sup > V:
            break

        func = ec_par if es_par else ec_impar

        try:
            f_inf = func(limite_inf)
            f_sup = func(limite_sup)

            if not np.isnan(f_inf) and not np.isnan(f_sup) and f_inf * f_sup < 0:
                u_solucion = brentq(func, limite_inf, limite_sup)

                # Cálculos de parámetros
                h = u_solucion * 2 / d
                w = np.sqrt(V**2 - u_solucion**2)
                q = w * 2 / d

                beta = np.sqrt((k0 * n2)**2 - h**2)
                n_eff = beta / k0

                tipo_modo = "par (simétrico)" if es_par else "impar (asimétrico)"

                modos.append({
                    'numero': m,
                    'tipo': 'par' if es_par else 'impar',
                    'u': u_solucion,
                    'w': w,
                    'h': h,
                    'q': q,
                    'beta': beta,
                    'n_eff': n_eff,
                    'k0': k0,
                    'V': V
                })

                print(f"TE{m:<4} {tipo_modo:<15} {u_solucion:<12.6f} {w:<12.6f} {h:<14.6f} {q:<14.6f} {beta:<14.6f} {n_eff:<10.6f}")

        except:
            continue

    return modos

# Resolver modos
modos = resolver_modos_guiados(n1, n2, d, wavelength, num_modos=4)

print(f"\n{len(modos)} Modos guiados TE")
```

### **7. VISUALIZACIÓN DE CAMPOS**

```python
print("\n" + "="*80)
print("VISUALIZACIÓN DE DISTRIBUCIONES DEL CAMPO ELÉCTRICO")
print("="*80)

# 4 subplots
fig, ejes = plt.subplots(2, 2, figsize=(14, 10))
ejes = ejes.flatten()

# Cálculos
x_calc = np.linspace(0, 5, 200)
z_calc = np.linspace(-1.2, d + 1.2, 250)
X_malla, Z_malla = np.meshgrid(x_calc, z_calc)

z_centro = d / 2

for indice, modo in enumerate(modos):
    if indice >= 4:
        break

    eje = ejes[indice]

    h = modo['h']
    q = modo['q']
    beta = modo['beta']
    es_par = (modo['tipo'] == 'par')
    num_modo = modo['numero']



    # Calcular campo
    E_campo = np.zeros_like(Z_malla)

    for i in range(Z_malla.shape[0]):
        for j in range(Z_malla.shape[1]):
            z_val = Z_malla[i, j]
            x_val = X_malla[i, j]

            # Perfil transversal
            if 0 <= z_val <= d:
                # Interior del núcleo
                if es_par:
                    amplitud = np.cos(h * (z_val - z_centro))
                else:
                    amplitud = np.sin(h * (z_val - z_centro))

            elif z_val < 0:
                # Cubierta inferior
                valor_frontera = np.cos(h * (-z_centro)) if es_par else np.sin(h * (-z_centro))
                amplitud = valor_frontera * np.exp(q * z_val)

            else:
                # Cubierta superior
                valor_frontera = np.cos(h * (d - z_centro)) if es_par else np.sin(h * (d - z_centro))
                amplitud = valor_frontera * np.exp(-q * (z_val - d))

            # Campo total: parte real
            E_campo[i, j] = amplitud * np.cos(beta * x_val)

    # Graficar
    mapa_color = eje.pcolormesh(X_malla, Z_malla, E_campo, shading='auto',
                                 cmap='RdBu_r', vmin=-1.0, vmax=1.0)

    # Interfaces
    eje.axhline(0, color='black', linewidth=2, linestyle='--', alpha=0.8)
    eje.axhline(d, color='black', linewidth=2, linestyle='--', alpha=0.8)

    # Título y etiquetas
    tipo_modo = "par" if es_par else "impar"
    eje.set_title(f'Modo TE$_{{{num_modo}}}$ ({tipo_modo})\n'
                  f'β = {beta:.4f} μm$^{{-1}}$, n$_{{eff}}$ = {modo["n_eff"]:.4f}',
                  fontsize=11, fontweight='bold')
    eje.set_xlabel('x (μm)', fontsize=10)
    eje.set_ylabel('z (μm)', fontsize=10)
    eje.set_xlim(0, 5)
    eje.set_ylim(-1.2, d + 1.2)

    # Barras de color
    plt.colorbar(mapa_color, ax=eje, label='Re[$E_y$]')



plt.suptitle(f'Campo Eléctrico: Modos TE en Guía Simétrica (n$_2$={n2}, n$_1$={n1}, d={d} μm)',
             fontsize=13, fontweight='bold', y=0.995)
plt.tight_layout()
plt.savefig('modos_TE_campo.png', dpi=150, bbox_inches='tight')
print("Gráfica DE Modos TE")
plt.show()
```

### **10. ECUACIÓN CARACTERÍSTICA**

```python
print("\n" + "="*80)
print("VISUALIZACIÓN DE LA ECUACIÓN CARACTERÍSTICA")
print("="*80)

fig, (eje1, eje2) = plt.subplots(1, 2, figsize=(14, 5))

V_valor = modos[0]['V'] if modos else 4.0
rango_u = np.linspace(1e-3, V_valor - 1e-3, 1000)

# Calcular funciones
lhs_par = []
lhs_impar = []

for u in rango_u:
    try:
        lhs_par.append(u * np.tan(u))
        lhs_impar.append(-u / np.tan(u))
    except:
        lhs_par.append(np.nan)
        lhs_impar.append(np.nan)

rhs = np.sqrt(V_valor**2 - rango_u**2)

# Gráfica 1: Ecuación característica
eje1.plot(rango_u, lhs_par, 'b-', linewidth=2, label='u·tan(u) [pares]')
eje1.plot(rango_u, lhs_impar, 'r-', linewidth=2, label='-u·cot(u) [impares]')
eje1.plot(rango_u, rhs, 'g--', linewidth=2.5, label=f'√(V²-u²), V={V_valor:.2f}')

# Marcar intersecciones
for modo in modos:
    u_sol = modo['u']
    w_sol = modo['w']
    color = 'blue' if modo['tipo'] == 'par' else 'red'
    marcador = 'o' if modo['tipo'] == 'par' else 's'
    eje1.plot(u_sol, w_sol, marcador, markersize=10, color=color, zorder=5)

eje1.set_xlabel('u = h·d/2', fontsize=11, fontweight='bold')
eje1.set_ylabel('w = q·d/2', fontsize=11, fontweight='bold')
eje1.set_title('Ecuación Característica para Modos TE', fontsize=12, fontweight='bold')
eje1.set_xlim(0, V_valor)
eje1.set_ylim(0, V_valor)
eje1.grid(True, alpha=0.3)
eje1.legend(fontsize=10)
eje1.set_aspect('equal')

# Gráfica 2: Error (ceros de la ecuación)
errores_par = []
errores_impar = []

for u in rango_u:
    try:
        w = np.sqrt(V_valor**2 - u**2)
        errores_par.append(u * np.tan(u) - w)
        errores_impar.append(-u / np.tan(u) - w)
    except:
        errores_par.append(np.nan)
        errores_impar.append(np.nan)

eje2.plot(rango_u, errores_par, 'b-', linewidth=2, label='Modos pares')
eje2.plot(rango_u, errores_impar, 'r-', linewidth=2, label='Modos impares')
eje2.axhline(0, color='k', linestyle='-', linewidth=0.8)

for modo in modos:
    eje2.plot(modo['u'], 0, 'o' if modo['tipo'] == 'par' else 's',
             markersize=10, color='blue' if modo['tipo'] == 'par' else 'red', zorder=5)

eje2.set_xlabel('u = h·d/2', fontsize=11, fontweight='bold')
eje2.set_ylabel('Error característico', fontsize=11, fontweight='bold')
eje2.set_title('Búsqueda de Raíces', fontsize=12, fontweight='bold')
eje2.grid(True, alpha=0.3)
eje2.legend(fontsize=10)

plt.tight_layout()
plt.savefig('ecuacion_caracteristica.png', dpi=150, bbox_inches='tight')

plt.show()
```
